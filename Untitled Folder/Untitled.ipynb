```python
import pandas as pd
import matplotlib.pyplot as plt
%matplotlib inline

import pmdarima as pmd
from prophet import Prophet
from prophet.plot import add_changepoints_to_plot
from statsmodels.tsa.holtwinters import ExponentialSmoothing
from sklearn.metrics import r2_score
```

```python
df = pd.read_csv('global-energy-substitution.csv', index_col='Year', parse_dates=True)
df.head()
```

```python
df[['Other_renewables', 'Biofuels', 'Solar', 'Wind', 'Hydropower','Nuclear', 'Gas', 'Oil', 'Coal', 'Traditional_biomass']].plot(figsize=(12,6));
```

```python
df.columns
```

```python
Gas = df['Gas'].copy()
```

```python
Gas = pd.DataFrame(Gas)
```

```python
Gas.head()
```

```python
Gas.shape
```

```python
Ntest = 10
train = Gas.iloc[:-Ntest]
test = Gas.iloc[-Ntest:]
```

```python
model = pmd.auto_arima(train, error_action='ignore', trace=True, supress_warning=True, max_p=12, max_q=2, max_order=14,
                     stepwise=False, seasonal=True)
```

```python
model.summary()
```

```python
test_pred, confint = model.predict(n_periods=Ntest, return_conf_int=True)
```

```python
train_pred = model.predict_in_sample(start=1, end=-1)

fig, ax = plt.subplots(figsize=(12,6))
ax.plot(Gas.index, Gas, label='data')
ax.plot(train.index[1:], train_pred, label='fitted')
ax.plot(test.index, test_pred, label='forecast')
ax.fill_between(test.index, \
               confint[:,0], confint[:,1], \
               color='red', alpha=0.3)
ax.legend();
```

# ESM

```python
train_idx = df.index <= train.index[-1]
test_idx = df.index > train.index[-1]
```

```python
hw = ExponentialSmoothing(
                        train,
                        trend='add',
                        seasonal=None,
                        initialization_method='heuristic',
                        use_boxcox=True
                        )
res_hw = hw.fit()
```

```python
Gas.loc[train_idx, 'HoltWinterTrain'] = res_hw.fittedvalues
```

```python
Gas.loc[test_idx, 'HoltWinterTest'] = res_hw.forecast(Ntest)
```

```python
Gas[['Gas','HoltWinterTrain','HoltWinterTest']].plot(figsize=(12,6));
```
