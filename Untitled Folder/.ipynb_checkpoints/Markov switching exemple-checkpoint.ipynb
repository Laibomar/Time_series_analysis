```python
import numpy as np
import statsmodels.api as sm
import matplotlib.pyplot as plt
```

```python
# Génération de données simulées
np.random.seed(0)
nobs = 100
beta = [1, -2]
states = np.zeros(nobs)
states[40:60] = 1  # Régime 1
states[70:90] = 2  # Régime 2
X = np.random.normal(size=(nobs, 2))
Y = np.dot(X, beta) + np.random.normal(size=(nobs))

# Modélisation de la série à changements de régime
mod = sm.tsa.MarkovRegression(endog=Y, exog=X, k_regimes=3)
res = mod.fit()

# Affichage des résultats
print(res.summary())
```

```python
# Plot de la série et des régimes
plt.figure(figsize=(12, 6))
plt.plot(Y, label='Série temporelle (PIB simulé)')
plt.plot(states, linestyle='--', label='Régimes prédits', color='red')
plt.xlabel('Temps')
plt.ylabel('Valeurs')
plt.title('Série temporelle avec Changements de Régime')
plt.legend()
plt.grid(True)
plt.show()
```

```python
# Afficher les statistiques de test pour les transitions de régime
print(res.regime_transition)
```
