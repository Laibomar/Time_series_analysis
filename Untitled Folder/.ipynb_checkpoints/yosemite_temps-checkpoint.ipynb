```python
import pandas as pd
import pmdarima as pmd
import seaborn as sns
import matplotlib.pyplot as plt
%matplotlib inline

from prophet import Prophet
from prophet.plot import add_changepoints_to_plot
from statsmodels.tsa.holtwinters import ExponentialSmoothing
from sklearn.metrics import r2_score
```

```python
df = pd.read_csv('example_yosemite_temps.csv')
df.head()
```

```python
df.tail(20)
```

```python
df['y'].isnull().sum()
```

```python
df['y'].fillna(df['y'].mean(), inplace=True)
```

```python
df['y'].isna().sum()
```

```python
df['y'].plot(figsize=(12,5));
```

```python
m = Prophet(seasonality_mode='multiplicative')
m.fit(df)
```

```python
future = m.make_future_dataframe(periods=30 , freq='H')
forecast = m.predict(future)
```

```python
m.plot(forecast);
```

```python
m.plot_components(forecast);
```

```python
y_pred = forecast['yhat']
```

```python
print(r2_score(df['y'],y_pred[:18721]))
```

```python
import numpy as np
```

```python
def rmse(t, y):
    return np.sqrt(np.mean((t - y)**2))
```

```python
print("RMSE:", rmse(df['y'], y_pred[:18721]))
```

# ESM

```python
df.columns
```

```python
df.set_index(df['ds'], inplace=True)
```

```python
df.index = pd.to_datetime(df.index).tz_localize(None)
```

```python
df.drop(['ds'],axis=1, inplace=True)
```

```python
df.head()
```

```python
df.plot(figsize=(12,5));
```

```python
df.index.freq = '5T'
df.index
```

```python
Ntest = 30
train = df.iloc[:-Ntest]
test = df.iloc[-Ntest:]
```

```python
train_idx = df.index <= train.index[-1]
test_idx = df.index > train.index[-1]
```

```python
hw = ExponentialSmoothing(train['y'],
                         initialization_method = 'legacy-heuristic',
                         trend='add', seasonal='add', seasonal_periods=24)

res_hw = hw.fit()
```

```python
df.loc[train_idx, 'HoltWinterTrain'] = res_hw.fittedvalues
```

```python
df.loc[test_idx, 'HoltWinterTest'] = res_hw.forecast(Ntest)
```

```python
df[['y','HoltWinterTrain','HoltWinterTest']].plot(figsize=(12,6));
```

```python
fig, ax = plt.subplots(figsize=(10,5))
ax.plot(test.index, test, label='data')
ax.plot(test.index, res_hw.forecast(Ntest), label='forecast')
ax.legend();
```
