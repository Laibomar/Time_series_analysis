```python
import pandas as pd
import matplotlib.pyplot as plt

# Exemple de données de prix de clôture du Bitcoin sur une période de 100 jours
data = {
    'Date': ['2023-01-01', '2023-01-02', '2023-01-03', '2023-01-04', '2023-01-05', '2023-01-06'],
    'BTC_Close_Price': [50000, 52000, 53000, 52500, 54000, 55000]
}

# Création d'un DataFrame à partir des données
df = pd.DataFrame(data)

# Convertir la colonne 'Date' en objet DateTime
df['Date'] = pd.to_datetime(df['Date'])

# Tri des données par date
df = df.sort_values('Date')

# Calcul de la moyenne mobile simple (SMA) sur 50 jours avec un décalage de 49 jours
sma_period = 50
df['50_Day_SMA'] = df['BTC_Close_Price'].rolling(window=sma_period, min_periods=1).mean()

# Affichage du DataFrame
print(df)

# Tracé du prix de clôture du BTC et de la SMA
plt.figure(figsize=(12, 6))
plt.plot(df['Date'], df['BTC_Close_Price'], label='Prix de clôture BTC', color='blue')
plt.plot(df['Date'], df['50_Day_SMA'], label='SMA 50 jours', color='red')
plt.xlabel('Date')
plt.ylabel('Prix')
plt.title('Moyenne Mobile Simple (SMA) sur 50 jours du Bitcoin')
plt.legend()
plt.grid()
plt.show()
```

```python
import yfinance as yf
import pandas as pd
import matplotlib.pyplot as plt

# Définissez les dates de début et de fin souhaitées
start_date = '2023-01-01'
end_date = '2023-10-30'  # Date actuelle, 30 octobre 2023

# Télécharger les données du BTC depuis Yahoo Finance pour la période spécifiée
btc = yf.download('BTC-USD', start=start_date, end=end_date)

# Calculer la moyenne mobile simple (SMA) sur 50 jours
sma_period = 50
btc['50_Day_SMA'] = btc['Close'].rolling(window=sma_period).mean()

# Tracé du prix de clôture du BTC et de la SMA
plt.figure(figsize=(12, 6))
plt.plot(btc.index, btc['Close'], label='Prix de clôture BTC', color='blue')
plt.plot(btc.index, btc['50_Day_SMA'], label='SMA 50 jours', color='red')
plt.xlabel('Date')
plt.ylabel('Prix')
plt.title('Moyenne Mobile Simple (SMA) sur 50 jours du Bitcoin')
plt.legend()
plt.grid()
plt.show()
```

```python
import yfinance as yf
import pandas as pd
import matplotlib.pyplot as plt

# Définissez les dates de début et de fin souhaitées
start_date = '2023-01-01'
end_date = '2023-10-30'  # Date actuelle, 30 octobre 2023

# Télécharger les données du BTC depuis Yahoo Finance pour la période spécifiée
btc = yf.download('BTC-USD', start=start_date, end=end_date)

# Calculer la moyenne mobile simple (SMA) sur 50 jours
sma_period = 50
btc['50_Day_SMA'] = btc['Close'].rolling(window=sma_period).mean()

# Calculer la moyenne mobile exponentielle (EMA) sur 50 jours
ema_period = 50
btc['50_Day_EMA'] = btc['Close'].ewm(span=ema_period, adjust=False).mean()

# Tracé du prix de clôture du BTC, de la SMA et de l'EMA
plt.figure(figsize=(12, 6))
plt.plot(btc.index, btc['Close'], label='Prix de clôture BTC', color='blue')
plt.plot(btc.index, btc['50_Day_SMA'], label='SMA 50 jours', color='red')
plt.plot(btc.index, btc['50_Day_EMA'], label='EMA 50 jours', color='green')
plt.xlabel('Date')
plt.ylabel('Prix')
plt.title('Moyenne Mobile Simple (SMA) et Exponentielle (EMA) sur 50 jours du Bitcoin')
plt.legend()
plt.grid()
plt.show()
```

```python
import yfinance as yf
import pandas as pd
import matplotlib.pyplot as plt

# Créez un objet Ticker pour le BTC
btc_ticker = yf.Ticker("BTC-USD")

# Obtenez les données historiques (prix de clôture) pour le BTC
btc_data = btc_ticker.history(period="1y")

# Calculer la moyenne mobile exponentielle (EMA) sur 50 jours
ema_period = 50
btc_data['50_Day_EMA'] = btc_data['Close'].ewm(span=ema_period, adjust=False).mean()

# Tracé du prix de clôture du BTC et de l'EMA
plt.figure(figsize=(12, 6))
plt.plot(btc_data.index, btc_data['Close'], label='Prix de clôture BTC', color='blue')
plt.plot(btc_data.index, btc_data['50_Day_EMA'], label='EMA 50 jours', color='green')
plt.xlabel('Date')
plt.ylabel('Prix')
plt.title('Moyenne Mobile Exponentielle (EMA) sur 50 jours du Bitcoin')
plt.legend()
plt.grid()
plt.show()
```

```python
import yfinance as yf
import mplfinance as mpf

# Créez un objet Ticker pour le BTC
btc_ticker = yf.Ticker("BTC-USD")

# Obtenez les données historiques (prix de clôture) pour le BTC avec des intervalles horaires
btc_data = btc_ticker.history(period="2d", interval="1h")  # Exemple : 2 jours de données avec des bougies horaires

# Calculer la moyenne mobile exponentielle (EMA) sur 50 heures
ema_period = 50
btc_data['EMA_50'] = btc_data['Close'].ewm(span=ema_period, adjust=False).mean()

# Tracé de l'EMA uniquement sans les bougies
mpf.plot(btc_data['EMA_50'], type='line', title="EMA 50 heures - BTC-USD", style='yahoo')
```

```python
import yfinance as yf
import pandas as pd
import mplfinance as mpf

# Créez un objet Ticker pour le BTC
btc_ticker = yf.Ticker("BTC-USD")

# Obtenez les données historiques (prix de clôture) pour le BTC avec des intervalles horaires
btc_data = btc_ticker.history(period="2d", interval="1h")  # Exemple : 2 jours de données avec des bougies horaires

# Calculer la moyenne mobile exponentielle (EMA) sur 50 heures
ema_period = 50
btc_data['EMA_50'] = btc_data['Close'].ewm(span=ema_period, adjust=False).mean()

# Tracé du prix de clôture du BTC avec l'EMA
kwargs = dict(type='candle', title="BTC-USD", mav=(50,), volume=True, style="yahoo")
mpf.plot(btc_data, **kwargs, addplot=[mpf.make_addplot(btc_data['EMA_50'], color='g', secondary_y=False, panel=1)])
```

```python
import yfinance as yf
import mplfinance as mpf

# Créez un objet Ticker pour le BTC
btc_ticker = yf.Ticker("BTC-USD")

# Obtenez les données historiques (prix de clôture) pour le BTC avec des intervalles horaires
btc_data = btc_ticker.history(period="2d", interval="1h")  # Exemple : 2 jours de données avec des bougies horaires

# Calculer la moyenne mobile exponentielle (EMA) sur 50 heures
ema_period = 50
btc_data['EMA_50'] = btc_data['Close'].ewm(span=ema_period, adjust=False).mean()

# Tracé du prix de clôture du BTC et de l'EMA
add_plot = [mpf.make_addplot(btc_data['EMA_50'], color='green', panel=1, secondary_y=False)]

mpf.plot(btc_data, type='candle', title="BTC-USD", volume=True, addplot=add_plot, style='yahoo')
```

```python
import yfinance as yf
import pandas as pd
import mplfinance as mpf

# Créez un objet Ticker pour le BTC
btc_ticker = yf.Ticker("BTC-USD")

# Obtenez les données historiques (prix de clôture) pour le BTC avec des intervalles horaires
btc_data = btc_ticker.history(period="2d", interval="1h")  # Exemple : 2 jours de données avec des bougies horaires

# Calculer la moyenne mobile exponentielle (EMA) sur 50 heures
ema_period = 50
btc_data['EMA_50'] = btc_data['Close'].ewm(span=ema_period, adjust=False).mean()

# Tracé du prix de clôture du BTC avec l'EMA
add_plot = [
    mpf.make_addplot(btc_data['EMA_50'], panel=1, secondary_y=False, color='green', linestyle='dashed')
]

mpf.plot(btc_data, type='candle', title="BTC-USD", volume=True, addplot=add_plot, style='yahoo')
```
