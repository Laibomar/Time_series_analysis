```python
from turtle import *

# Fonction pour dessiner un rectangle (carrosserie)
def draw_rectangle(width, height, color):
    fillcolor(color)
    begin_fill()
    for _ in range(2):
        forward(width)
        left(90)
        forward(height)
        left(90)
    end_fill()

# Fonction pour dessiner des roues
def draw_wheel(radius, color):
    fillcolor(color)
    begin_fill()
    circle(radius)
    end_fill()

# Dessiner la carrosserie de la voiture
draw_rectangle(200, 50, 'blue')

# Positionner pour dessiner la première roue
penup()
goto(-80, -50)
pendown()

# Dessiner la première roue
draw_wheel(20, 'black')

# Positionner pour dessiner la deuxième roue
penup()
goto(80, -50)
pendown()

# Dessiner la deuxième roue
draw_wheel(20, 'black')

# Cacher la tortue à la fin
hideturtle()

# Fermer la fenêtre lors d'un clic
exitonclick()
```
